```python
# Count how many times each invoice_id appears
id_counts = invoices['invoice_id'].value_counts()

# Keep only the ones that appear more than once — these are the "duplicate groups"
duplicate_groups = id_counts[id_counts > 1]

print("Number of duplicate groups:", len(duplicate_groups))
print("Total duplicate rows:", duplicate_groups.sum())
```

```python
# This shows HOW MANY groups have exactly 2 rows, exactly 3 rows, etc.
breakdown = duplicate_groups.value_counts().sort_index()
print(breakdown)
```

```python
# Find the invoice_id(s) with more than 2 duplicates
three_plus = duplicate_groups[duplicate_groups > 2]
print(three_plus)
```

```python
# Step 1: Count duplicate groups for payment_id
pid_counts = payments['payment_id'].value_counts()
duplicate_pid_groups = pid_counts[pid_counts > 1]

print("Number of duplicate groups:", len(duplicate_pid_groups))
print("Total duplicate rows:", duplicate_pid_groups.sum())

# Step 2: Breakdown by group size
breakdown_pid = duplicate_pid_groups.value_counts().sort_index()
print(breakdown_pid)
```

```python
# Merge payments with invoices to compare payment_date vs invoice_date
merged = payments.merge(invoices[['invoice_id', 'invoice_date', 'customer_id', 'branch_id']], on='invoice_id', how='left')

# Filter to only the payments that happened before their invoice date
early_payments = merged[merged['payment_date'] < merged['invoice_date']]

print("Total early payments found:", len(early_payments))
print()
print("By branch:")
print(early_payments['branch_id'].value_counts())
print()
print("By customer (top 10):")
print(early_payments['customer_id'].value_counts().head(10))
```
