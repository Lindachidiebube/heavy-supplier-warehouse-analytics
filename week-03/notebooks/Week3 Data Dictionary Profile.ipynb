```python
import pandas as pd, glob, os

BASE_PATH = '/content/drive/MyDrive/Cadetx Project/HeavySuppliersWarehouseDatasets/'
files = sorted(glob.glob(BASE_PATH + '*.csv'))
print("CSV files found:", len(files), "(we expect 12)")

out = []
for f in files:
    df = pd.read_csv(f)
    out.append(f"\n=== TABLE: {os.path.basename(f)} | rows: {len(df)} | columns: {len(df.columns)}")
    for c in df.columns:
        s = df[c]
        line = f"- {c} | type: {s.dtype} | empty: {int(s.isna().sum())} | distinct: {s.nunique()}"
        if pd.api.types.is_numeric_dtype(s):
            line += f" | min: {s.min()} | max: {s.max()}"
        else:
            vals = s.dropna().astype(str)
            if len(vals):
                line += f" | first: {vals.min()} | last: {vals.max()}"
        line += " | samples: " + "; ".join(s.dropna().astype(str).unique()[:5])
        out.append(line)

with open('/content/data_profile.txt', 'w') as fh:
    fh.write("\n".join(out))
print("\n".join(out)[:1500])
print("\nSaved: /content/data_profile.txt")
```
